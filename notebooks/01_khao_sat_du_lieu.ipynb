# 01. Khảo sát dữ liệu

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import os

# Đường dẫn trỏ tới file reviews.csv ở data/
DATA_PATH = os.path.join("..", "data", "reviews.csv")
# Nếu file nằm trong data/raw/reviews.csv thì dùng dòng dưới:
# DATA_PATH = os.path.join("..", "data", "raw", "reviews.csv")

df = pd.read_csv(DATA_PATH)

print(f"Tổng số dòng ban đầu: {len(df)}")
df.head()
```

Kiểm tra cấu trúc cột và giá trị thiếu (Null/NaN)

```python
# Xem thông tin kiểu dữ liệu và giá trị thiếu
print("--- THÔNG TIN CÁC CỘT ---")
df.info()

print("\n--- SỐ LƯỢNG GIÁ TRỊ THIẾU ---")
print(df.isnull().sum())
```

 Kiểm tra phân bố số sao (Rating)

```python
# Xác định tên cột rating (thường là 'rating' hoặc 'rating_star')
rating_col = [col for col in df.columns if 'rating' in col.lower() or 'star' in col.lower()]
rating_col = rating_col[0] if rating_col else None

if rating_col:
    print(f"Phân bố theo cột {rating_col}:")
    print(df[rating_col].value_counts().sort_index())
    
    # Biểu đồ phân bố
    plt.figure(figsize=(6, 4))
    sns.countplot(data=df, x=rating_col, palette="viridis")
    plt.title("Phân bố số sao đánh giá (Rating)")
    plt.xlabel("Số sao")
    plt.ylabel("Số lượng")
    plt.show()
else:
    print("Không tìm thấy cột rating, kiểm tra các cột sau:", df.columns.tolist())
```

Quy đổi số sao sang nhãn cảm xúc theo quy ước đặc tả

```python
def map_sentiment(val):
    if val in [1, 2]:
        return 'negative'
    elif val == 3:
        return 'neutral'
    elif val in [4, 5]:
        return 'positive'
    return 'unknown'

if rating_col:
    df['sentiment'] = df[rating_col].apply(map_sentiment)
    print("--- PHÂN BỐ NHÃN CẢM XÚC QUY ĐỔI ---")
    print(df['sentiment'].value_counts(normalize=True) * 100)
    
    plt.figure(figsize=(6, 4))
    sns.countplot(data=df, x='sentiment', order=['positive', 'neutral', 'negative'], palette="Set2")
    plt.title("Phân bố nhãn cảm xúc (Positive / Neutral / Negative)")
    plt.show()
```

Phân tích độ dài văn bản đánh giá

```python
# Tìm cột chứa văn bản (thường là 'comment', 'content', 'review_text')
text_col = [col for col in df.columns if col.lower() in ['comment', 'content', 'review', 'text']]
text_col = text_col[0] if text_col else None

if text_col:
    # Tính số từ trong mỗi bình luận
    df['word_count'] = df[text_col].astype(str).apply(lambda x: len(x.split()))
    
    print("--- THỐNG KÊ ĐỘ DÀI (SỐ TỪ) ---")
    print(df['word_count'].describe(percentiles=[0.5, 0.75, 0.9, 0.95, 0.99]))
    
    plt.figure(figsize=(8, 4))
    sns.histplot(df['word_count'], bins=50, kde=True, color='purple')
    plt.title("Phân bố độ dài bình luận (số từ)")
    plt.xlabel("Số từ")
    plt.xlim(0, 150)
    plt.show()
```
